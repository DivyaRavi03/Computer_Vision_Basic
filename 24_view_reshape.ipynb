## View() and Reshape()

- Both helps to change the shape of tensors
- Key differences:
  View() can be used only if tensor is stored in contiguous memory.
  Reshape() - can handle non-contiguous data and tensors, can also create a new copy of tensor if non-c.m
- Contiguous memory - if the original tensor has 2 x 3 matrix, the memory must store it has flattened array, must row it has row major order ( left to right, row by row), performs well
- Non-Contiguous memory - data remains same in memory but the access pattern changes, where indexing becomes computationally expensive

```python
import torch
```

```python
tensor = torch.arange(12)
```

```python
view_tensor = tensor.view(2,6)
reshape_tensor = tensor.reshape(2,6)

print(view_tensor)
print(reshape_tensor)
```

```python
# to check whether data is contiguous
tensor.is_contiguous()
```

```python
# when we are not sure about the number of cols needed, mention -1
infered_shape = tensor.view(3,-1) # pytorch will calculate number of cols needed based on number of rows mentioned
print(infered_shape)
```

```python
tensor_3d = tensor.reshape(2,3,-1)
print(tensor_3d)
```

```python
tensor_3d.is_contiguous()
```

```python
print(reshape_tensor)
transpose_tensor = reshape_tensor.t()
print(transpose_tensor)
```

```python
transpose_tensor.is_contiguous()
```

```python
# as it is non_contiguous view() will show an error
error = transpose_tensor.view(2,6)
```

```python
error = transpose_tensor.reshape(2,6)
print(error)
```

```python
# to make data contiguous
contiguous_tensor = transpose_tensor.contiguous()
error = contiguous_tensor.view(6,2)
print(error)
```
