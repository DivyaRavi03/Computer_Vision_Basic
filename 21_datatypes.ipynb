Tensor DATA TYPES

Base Data Types-
* float - for most deep learning tasks --> memory consumption raises for float16, float 32, float 64. lower precision, faster the speed but float64 is more precise than float32.
* integer - categorical data or indicies
* boolean - mask or logical operations
* complex numbers - for advanced computation

```python
import torch
```

```python
default_tensor = torch.tensor([1.5, 2.5, 3.5]) # default datatype for pytorch tensor is float32
print(default_tensor.dtype)
```

```python
float_tensor = torch.tensor([1.5, 2.5, 3.5], dtype=torch.float64)
print(float_tensor.dtype)
```

```python
int_tensor = torch.tensor([1, 2, 3], dtype=torch.int32) # int32 = general purpose integer
print(int_tensor.dtype)
print(int_tensor)
```

```python
bol_tensor = torch.tensor([True, False, True]) # takes this as boolean automatically, can also be explicitly mentioned using dtype = " "
print(bol_tensor.dtype)
```

```python
# to convert from 1 datatype to another (float to int)
float_tensor = torch.tensor([1.5, 2.5, 3.5])
print(float_tensor.dtype)
print(float_tensor)

int_tensor = float_tensor.to(torch.int64) # int64 = tensor indices
print(int_tensor.dtype)
print(int_tensor)
```

```python
# int to bool
int_tensor = torch.tensor([0,1,2,0,-1])
bool_tensor = int_tensor.to(torch.bool)
print(bool_tensor.dtype)
print(bool_tensor)
```

Impact of datatype on memory

```python
float32_tensor = torch.ones(10, dtype=torch.float32) # 32 - bit floating points
float64_tensor = torch.ones(10, dtype=torch.float64) # 64 - bit high precision computation

# tensor.element_size() - gives the size of one element in bytes
# to get total number of memory consumption for the tensor

print(float32_tensor.element_size())
print(float32_tensor.nelement())
print("Memory used by float 32 tensor: ", float32_tensor.element_size()*float32_tensor.nelement(), "bytes") # nelement gives total number of elements in tensor
print(float64_tensor.element_size())
print(float64_tensor.nelement())
print("Memory used by float 64 tensor: ", float64_tensor.element_size()*float64_tensor.nelement(), "bytes")
```
