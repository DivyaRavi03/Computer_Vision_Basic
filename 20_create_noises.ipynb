```python
import torch
```

Generate random numbers

```python
random = torch.rand(5) # prints 5 random value between 0 and 1
print(random)
print(random.ndim)
```

```python
# to create 3D matrix
torch.manual_seed(42) # seed, to have the same set of numbers every time we run it eventhough rand() is defined
random_matrix = torch.rand(2,5,3) # 2 of 5 rows and 3 cols
print(random_matrix)
print(random_matrix.ndim)
print(random_matrix.shape)
```

Generate random integers
Syntax - torch.randint(low,high,size)

```python
random_int = torch.randint(0,11,(5,5))
print(random_int)
```

```python
# generate random numbers from normal distribution. torch.normal(mean,std, size)
random_normal = torch.normal(0.5, 2, (5,))
print(random_normal)
```

Creating noise image using pytorch

```python
import matplotlib.pyplot as plt

# h, w of image (256, 256)

uniform_noise = torch.rand((256, 256))
plt.imshow(uniform_noise, cmap='gray')
plt.title('Uniform Noise Image')
plt.axis("off")
plt.show()
```

```python
# generating an image with rgb values
# h, w, c of an image 256 x 256 x 3
uniform_noise_rgb = torch.rand((256, 256, 3))
plt.imshow(uniform_noise_rgb.numpy()) # most of lib like opencv2, matplotlib runs well in numpy more than torch as there will be no conversion error
plt.title('Uniform Noise RGB Image')
plt.axis("off")
plt.show()
```

```python
print("Shape:",uniform_noise_rgb.shape)
print("Data type:",uniform_noise_rgb.dtype)
print("Min value: ",uniform_noise_rgb.min()) # else torch.min(uniform_noise_rgb)
print("Max value: ",uniform_noise_rgb.max())
```

Tensors of Zero's and One's. These tensors are backbone of initializing models, placeholders or testing data in pytorch

Zero's

```python
zeros_vector = torch.zeros(5)
print("Vector: \n", zeros_vector)
print(zeros_vector.ndim)
print(zeros_vector.shape)

zeros_matrix = torch.zeros(10,10)
print("Matrix:\n",zeros_matrix)
print(zeros_matrix.ndim)
print(zeros_matrix.shape)
```

```python
plt.imshow(zeros_matrix.numpy(), cmap="gray",vmin=0,vmax=1)
plt.title("Dark image")
plt.axis("off")
plt.show()
```

One's

```python
ones_tensors = torch.ones(5)
print(ones_tensors)
print(zeros_matrix.ndim)
print(zeros_matrix.shape)

ones_matrix = torch.ones(10,10)
print("Matrix:\n",ones_matrix)
print(ones_matrix.ndim)
print(ones_matrix.shape)
```

```python
plt.imshow(ones_matrix.numpy(), cmap="gray",vmin=0,vmax=1)
plt.title("White image")
plt.axis("off")
plt.show()
```
