Creating tensors

Scalar Tensor - A tensor with zero dimension

```python
scalar = torch.tensor(42)
print("Scalar tensor: ", scalar)
print("Scalar dimension: ", scalar.ndim)
print("Scalar shape: ", scalar.shape)
```

Vector Tensor - 1 dimension tensor

```python
vector = torch.tensor([1,2,3])
print("Vector tensor: ", vector)
print("Vector dimension: ", vector.ndim)
print("Vector shape: ", vector.shape)
```

Matrix Tensor - 2 Dimension

```python
matrix = torch.tensor([[1,2],[3,4]])
print("Matrix tensor: ", matrix)
print("Matrix dimension: ", matrix.ndim)
print("Matrix shape: ", matrix.shape)
```

Tensor 3 D - more than 2 dimensional

```python
tensor_3d = torch.tensor([
                      [[1,2],[3,4]],
                      [[1,2],[3,4]]
                      ])
print("3D_tensor tensor: ", tensor_3d)
print("3D_tensor dimension: ", tensor_3d.ndim)
print("3D_tensor shape: ", tensor_3d.shape)
```

GPU accerlation

```python
if torch.cuda.is_available():
  gpu_tensor = vector.to('cuda') # moving vector to cuda
  print("Tensor on GPU", gpu_tensor)
else:
  print("Tensor on CPU as CUDA is not available on this machine")
```

Tensor Properties

```python
print("Data type of matrix: ", matrix.dtype)
print("Is matrix on GPU: ", matrix.is_cuda)
print("Matrix avaliable on CUDA: ",matrix.to('cuda'))
print("Matrix size: ", matrix.size())
print("Number of elements in matrix: ", matrix.numel())
```

Indexing tensors - selecting specific elements or range of elements from tensor.
Syntax: tensor[row, col]

```python
# Accessing elements in a tensor

tensor = torch.tensor([[10,20,30],[40,50,60],[70,80,90]])
print("Dimensions: ", tensor.ndim)
print("Shape: ", tensor.shape)
```

```python
print("Element at (0,1): ",tensor[0,1]) # scalar value
```

```python
# 0th row only
tensor[0]
```

```python
# all rows but only 0th col
tensor[:,0]
```

Slicing

```python
print("First 2 rows: ", tensor[:2])
```

```python
print("First 2 cols:\n ", tensor[:, :2])
```

```python
print("Middle element:\n ",tensor[1:2, 1:])
```

```python
# Fancy indexing 20=(0,1); 90 = (2,2)
print("Get elements 20 and 90: \n", tensor[[0,2],[1,2]]) # [[rows of 2 elements],[corresponding cols]]
```

```python
# Boolean indexing
mask = tensor > 50
print(mask)

print(tensor[mask]) # values
```

```python
# changing particular values via indexing
tensor[:,1] = 25
print(tensor)
```

```python
tensor[:,0] = torch.tensor([100,400,700])
print(tensor)
```

```python
# advanced indexing
indices = torch.tensor([0,2])
print(torch.index_select(tensor, dim = 0, index=indices)) # dim = 0 -> rows, 1 -> cols
print(torch.index_select(tensor, dim = 1, index=indices)) # dim = 1 -> cols
```

```python
# using tensor with steps
print(tensor[:, ::2])
```

```python
# reverse an array
print(torch.flip(tensor, dims=[0]))
print(torch.flip(tensor, dims=[1]))
```
