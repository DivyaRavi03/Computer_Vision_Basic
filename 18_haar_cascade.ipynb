# Haar Cascasde

```python
import cv2
```

## Face detection 

```python
cam = cv2.VideoCapture(0) # to use webcam

face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

while True:
    ret, frame = cam.read() # read/retireve frames from camera

    if not ret:
        print("Failed to grab frame") # if no frames
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.3, minNeighbors=5) # scaleFactor - helps to detect at different scale
    # minNeighbors - helps to reduce false postivies, higher the number, higher the computation, but too strict can even miss face

    for (x,y,w,h) in faces:
        cv2.rectangle(frame, (x,y), (x+w, y+h), (255,0,0), 2)
        cv2.putText(frame, "face detected", (x,y-10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0,255,0),2, cv2.LINE_AA)

    
    cv2.imshow('camera', frame)


    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cam.release()

cv2.destroyAllWindows
```

## Face + eye + mouth detection

To detect eye and mouth, 
1. first crop from whole image to face detected region 
2. convert it to grayscale image

```python
cam = cv2.VideoCapture(0) # to use webcam

face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
eye_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_eye.xml')
mouth_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_smile.xml')

while True:
    ret, frame = cam.read() # read/retireve frames from camera

    if not ret:
        print("Failed to grab frame") # if no frames
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.3, minNeighbors=5) # scaleFactor - helps to detect at different scale
    # minNeighbors - helps to reduce false postivies, higher the number, higher the computation, but too strict can even miss face

    for (x,y,w,h) in faces:

        # boundary box for face
        cv2.rectangle(frame, (x,y), (x+w, y+h), (255,0,0), 2)
        cv2.putText(frame, "face detected", (x,y-10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0,255,0),2, cv2.LINE_AA)

        # cropped region
        roi_gray = gray[y:y+h, x:x+w]     # extract grey part from the image
        roi_color = frame[y:y+h, x:x+w]   

        eyes = eye_cascade.detectMultiScale(roi_gray, scaleFactor=1.3, minNeighbors=5)
        mouth = mouth_cascade.detectMultiScale(roi_gray, scaleFactor=1.8, minNeighbors=10)

        for (ex,ey, ew, eh) in eyes: # points relative to main image
            cv2.rectangle(frame, (x + ex, y + ey), (x+ex+ew, y+ey+eh), (255,0,0), 2)
            cv2.putText(frame, "Eye", (x + ex, (y + ey)-10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0,255,0),2, cv2.LINE_AA)

        for (mx, my, mw, mh) in mouth:
            cv2.rectangle(frame, (x + mx,y + my), (x+mx+mw, y+my+mh), (255,0,0), 2)
            cv2.putText(frame, "Mouth", (x + mx, (y + my)-10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0,255,0),2, cv2.LINE_AA)

    
    cv2.imshow('camera', frame)


    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cam.release()

cv2.destroyAllWindows
```
